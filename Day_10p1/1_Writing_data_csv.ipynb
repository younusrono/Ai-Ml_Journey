# grouping and aggregation
Pandas
Grouping & Aggregation of Data
• groupby( )
• mean / sum / max / min / count etc.
• agg( )


```python
# calculate income mean
df.groupby("country")["income"].mean()

# calculate income min
df.groupby("country")["income"].min()

# calculate income max
df.groupby("country")["income"].max()

# gender income mean
df.groupby("gender")["income"].mean()

# agg is used to complete 3 function
df.groupby("country")["income"].agg(["mean","min","max"])

# rename the columns name
df.groupby("country")["income"].agg(avg_salary="mean",min_salary="min",max_salary="max")

# aaply agg function in different columns
df.groupby("country").agg({
    "income":"max",
    "age":"mean"
})

# if we want to change specific col name
df.groupby("country").agg(
   max_salary=("income","max"),
   mean_age=("age","mean")
)
```
