# Handle missing values


```python
import pandas as pd
```

```python
df=pd.read_csv("raw_data.csv")

# use to find the null values
df.isnull()
# nan thkhle ture lekha

# total number of values null
df.isnull().sum()

# use to drop the null values
df.dropna()

# to drop the columns
df.dropna(axis=1)

# use to fill up values with particular one
# df.fillna(10)
cleaned_data=df.copy()
age_mean=cleaned_data["age"].mean()
cleaned_data["age"]=cleaned_data["age"].fillna(age_mean)
cleaned_data


# forward fill
df.ffill()

# backward fill
df.bfill()
```
