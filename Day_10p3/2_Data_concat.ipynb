# data concatentation
# it can be done by row and column 


```python
import pandas as pd
```

```python
# Concatenation

df1 = pd.DataFrame({
    "id": [1, 2, 3],
    "name": ["Adam", "Eve", "Bob"]
})

df2 = pd.DataFrame({
    "id": [4, 5, 6],
    "name": ["Charlie", "Fester", "Gilian"]
})

# row wise concat
pd.concat([df1,df2])
pd.concat([df1,df2],ignore_index=True)

# column wise
pd.concat([df1,df2],axis=1)
```
