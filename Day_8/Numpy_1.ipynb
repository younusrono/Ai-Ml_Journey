```python
import numpy as np
```

```python
arr=np.array([1,2,3,4,5])
print(arr,type(arr),arr.dtype,arr.shape)
arr_2d=np.array([[1,2,3] ,[5,6,7],[9,10,11]])
print(arr_2d,arr_2d.shape)
```

```python
arr1=np.zeros((2,3),dtype="int64")
print(arr1,arr1.shape)
arr_2=np.ones((5,))
print(arr_2,arr_2.shape)
arr_3=np.full((3,3),100)
print(arr_3,arr_3.shape)
arr_4=np.eye((4))
print(arr_4,arr_4.shape)

arr_5=np.arange(1,10,1)
print(arr_5)
arr_6=np.linspace(1,100,4)
print(arr_6)
```

```python
arr=np.array([1,2,3,4,5])
print(arr.shape)
print(arr.size)
print(arr.ndim)
print(arr.dtype)
arr_1=arr.astype(np.float64)
print(arr_1,arr_1.dtype)
```

```python
# operation of array ---
# 1) reshaping
arr=np.array([[1,2,3],[2,3,4]])
print(arr,arr.shape)
re_arr=arr.reshape((3,2))
print(re_arr,re_arr.shape)

flattend=arr.flatten()
print(flattend,flattend.shape)
```

```python
# indexing
arr=np.array([1,2,3,4,5])
print(arr[0])
arr_1=np.array([[1,2],[3,4],[6,7]])
print(arr_1[1,1])
```

```python
# fancy Indexing
arr=np.array([[1,2],[3,4],[6,7]])
idx=[2,0],[0,0]
print(arr[idx])
```

```python
# boolean indexing
arr=np.array([1,2,3,4,5,6])
print(arr[arr>3])
```
