```python
import json
```

```python
# load data
def load_data(filename):
    with open (filename,"r") as f:
        data=json.load(f)
    return data
```

```python
data=load_data("store_data.json")
```

```python
# Clean & structure the data
def clean_data(data):
    text_to_num = {"one": 1, "two": 2, "three": 3, "four": 4, "five": 5}
    cleaned_data = []
    unique_users = set()

    for user in data:

        # Clean ratings - data consistency
        raw_rating = str(user["rating"]).strip().lower()
        if(raw_rating in text_to_num):
            raw_rating = text_to_num[raw_rating]
        user["rating"] = raw_rating

        # Handle missing vals
        raw_age = user.get("age")
        if(raw_age == None):
            user["age"] = None

        # Deduplication
        if(user["name"].strip() in unique_users):
            continue
        unique_users.add(user["name"])
        cleaned_data.append(user)

    return cleaned_data
 
```

```python
data=clean_data(data)
```

```python
# get meaning full insights of Data
def get_insight(data):
    total_rating=0
    for user in data :
        total_rating += float(user["rating"])
    print(f"avg of rating is {total_rating/len(data)}")

    # the parcentage of user poor rating
    poor_rating=0
    for user in data:
        if(float(user["rating"])<3):
            poor_rating+=1

    print(f"poor rating parcentage is : {poor_rating/len(data) *100}%")
    
```

```python
get_insight(data)
```

```python
# build recomanded features
# 1. If user rating >= 4, recommend same brand products.
# 2. If user rating < 4, recommend different brand products.

#Build recommendation Feature
def get_recommendations(data):
    recommendations = []
    
    for user in data:
        curr_recomm = {}
        curr_recomm["name"] = user["name"]
        
        if(float(user['rating']) >= 4):
            curr_recomm["brand"] = "Apple"
        else:
            curr_recomm["brand"] = "Samsung"
            
        recommendations.append(curr_recomm)
    return recommendations

```

```python
get_recommendations(data)
```
