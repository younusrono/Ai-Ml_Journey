```python
# Numpy Vectorization and Boradcasting
```

```python
import numpy as np
```

```python
arr_1=np.array([1,2,3,4,5])
arr_2=np.array([6,7,8,9,10])
arr=arr_1+arr_2
print(arr)
```

```python
#boardcasting
arr_1=np.array([1,2,3,4,5])
arr_2=np.array([[6,7,8,9,10],[6,7,8,9,10]])
print(arr_1+arr_2)
```

```python
# Normalization
arr=np.array([[1,2],[3,4]])
mean_val=(np.mean(arr))
Std_val=(np.std(arr))
Normalization_arr=(arr-mean_val)/Std_val
print(Normalization_arr)
print(mean_val)
print(Std_val)
```
